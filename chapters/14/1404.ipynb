## 14.4 射影と法ベクトル

::::{admonition} ask ai
:class: tip

* 平面に描かれた曲線の位置ベクトルの微分は接ベクトルです

:::{admonition} ask ai repeatedly
:class: seealso

* 接ベクトルは接空間の基底ベクトルです
* 曲線の接ベクトルの自身の内積の平方根は曲線の長さに関連します。
  積分すると曲線の長さが得られます
* 曲線の接ベクトルと単位法ベクトルとの内積は、曲線の曲率に関する情報を持ちます
* 接ベクトルの微分は必ずしも接ベクトルとは直交しません
  * 単位接ベクトルの微分は接ベクトルと直交する法線ベクトルですが、
    数式が複雑化する傾向があります
  * 接ベクトルの微分から、接ベクトルへの射影ベクトルを引くと
    法線方向の成分が残り法線ベクトルとなります
  * 接ベクトルの要素 $(x,y)$ を $(-y, x)$ に入れ替えると内積が$0$となり、法線ベクトルをもっとも簡単に求められます

## 13.2 円、楕円、球面

::::{admonition} ask ai
:class: tip

* 円を極座標で媒介変数表示したとき、位置ベクトルの微分は接ベクトルとなり、
  接ベクトルの微分は位置ベクトルと反対方向のベクトルとなります。
  それらは全て直交します

:::{admonition} ask ai repeatedly
:class: seealso

* 円の場合、接ベクトルの微分は接ベクトルへの射影を持ちません。
  このため、接空間である接線では、円が平坦なのか定曲率なのかを区別できません
* 楕円の接ベクトルの微分は接空間への射影を持ちます。
  接ベクトルの周期的な変化を観察すると楕円の長径と短径を特定できます
* 球面の極座標表示は、経線方向の大円と緯線方向の直交する円によって媒介変数表示されます
  * 経線方向の移動では、円の場合と同じく接ベクトルの変化は接空間に射影を持ちません
  * 緯線方向の移動では、緯線による切断面の円の中心は球面の軸上にあり、円の場合と同じく接ベクトルの微分は接線に射影を持ちません。
    しかし、切断面の円の中心が球面の中心と一致しない場合は、接空間に射影が生じます
  * 球面上の移動方向が緯線方向と経線方向の線型結合で表される場合、そのベクトルの微分は接空間に射影を持ちます
  * この性質から、円の場合とは異なり、球面の接空間は曲率に関する情報を内在しています
:::
