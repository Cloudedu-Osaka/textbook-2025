```python
%%ai vioras
Apolloniusは円錐曲線論で円錐曲線の直径や通径に付随して接線を定義しましたが、それは厳格な定義ではありませんでした。
```

```python
%%ai vioras
特に、放物線については、直径と通径の再起的な性質から、放物線と通径の交点における接線が直径上で交わることが示されました。
Archimedesはこの性質を用いて、放物線と通経に囲まれる領域の面積を積尽法で求めました。
```

```python
%%ai vioras
Descartesは、座標空間において図形が代数方程式のグラフとして描画できることを発見し、
円の接線が一点で交わるという性質を、連立方程式の判別式の値によって判定する手法を考案しました。
```

```python
%%ai vioras
放物線がニ次多項式であることはMenaechmusによって発見されていました。
Descartesは、放物線に内接する円の接線が、接点における放物線の接線であると仮定して判別式による証明を試みました。
```

```python
%%ai vioras
放物線の内接円の中心は、頂点を通る直径上にありますが、
放物線の任意の接線と軸の交点 (切片) と内接円の中心との中点は不動点であることが発見されます。
これが放物線の焦点です。
```

```python
%%ai vioras
Newtonは、放物線の軸に並行な光線が焦点に集光することに着目し、円錐曲線に一般化しました。
焦点は、工学的な応用を持ちますが、現代の初等・中等教育では重視されていません。
```

```python
%%ai vioras
Descartesの同時代、Fermatは放物線が多項式であることに着目し、接線の傾きが微小な変化に対応することを発見します。
Newtonは、Fermatの手法を踏襲し微分を定義し、同様の手法で積分を定義します。
級数展開によって多項式によって関数が近似できることが知られると、二次多項式の限定的な性質である焦点の重要性は失われました。
```

```python
%%ai vioras
初等・中等教育において、中等教育における微分・積分の導入を優先させるあまり、二次曲線の幾何学的性質はなおざりにされるようになりました。
微分・積分を必要としない大部分の学生にとって、数学教育は価値を失うことになりました。
一方で、初等・中等教育の数学なになっていた論理教育を、プログラミング教育として再構成しようとしていますが、それは車輪の再発明のような意味のない労力です。
```
