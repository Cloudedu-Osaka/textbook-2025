## 12.3 FermatとNewtonと微分・積分 

::::{admonition} ask ai
:class: tip

* Fermatは、多項式の変数 $x$ を微少量を含む $x + dx$ に置き換えて展開することで、
  多項式の接線が二項係数に関連することを示しました

:::{admonition} ask ai repeatedly
:class: seealso

* Newtonは、Fermatの発見がそのまま多項式の積分に応用できることを発見し、微分・積分を流率 fluxion として体系化しました
* Newtonは、Fermatの二項展開の応用を検討する過程で二項係数の一般化に成功しました
* 円の方程式 $\sqrt(1-x^2)$ の多項式による近似方法を発見したNewtonは、
  古代インドで発見され、イスラム経由で伝わった三角関数の級数展開を理論化しました
* Newtonは、$\sqrt(1-x^2)$の積分から$\arcsin$の級数展開式を求め、
  逆関数の関係式によって$\sin$の級数展開を求めました
