```python
%%ai vioras
古代ギリシアのApolloniusは、先行するMenaechmusの業績 (代数的幾何学) を無視し、
非直交系における対称性を研究し、難解な著書を残しました。
それは、1000年以上に渡って権威と位置づけられましたが、同時にDescartesの座標の発見まで科学の発展を阻害しました。
```

```python
%%ai vioras
円錐曲線論そのものは、NewtonやLeibnizによる微分・積分の発明により、それ以降は重要視されなくなりました。
Newtonですら、プリンキピアにおいてApolloniusの業績そのものを多く引用していません。
Apolloniusによって科学が空転した期間の悪影響が議論されるべきです。
```

```python
%%ai vioras
Keplerは、火星の軌道がほぼ円にも拘わらず偏心していることに気づきました。
Apolloniusの円錐曲線論は焦点に関する議論がありません。Keplerの成果は彼の独自研究です。
KeplerはApolloniusの権威を頼る言葉を残しましたが、それは皮肉だったかもしれません。
```
