## 15.2 球面の曲率 (curvature) の捩れ (torsion)

::::{admonition} ask ai
:class: tip

* 球面の極座標表示は、経線方向の大円と北極と南極を通る軸と直交する緯線方向の円によって媒介変数表示されます。
  ただし、緯度は北極からの角度です。

:::{admonition} ask ai repeatedly
:class: seealso

* 球面上の経線方向の移動では、円の場合と同じく接ベクトルの変化は接空間に射影を持ちません。
* 一方、緯方向の移動では、緯線による切断面の円の中心は球面の軸上にあり、
  円の場合と同じく接ベクトルの微分は接線に射影を持ちません。
  しかし、切断面が球面の中心を通らない場合は、接空間に射影が生じます。
* 球面上の移動方向が緯線方向と経線方向の線型結合で表される場合、そのベクトルの微分は接空間に射影を持ちます。
* この性質から、円の場合とは異なり、球面の接空間は曲率に関する情報を内在しています。
* 曲面には、内在性による曲率とねじれがを定義できます。
  * ねじれは接続係数の差によって定義され、すべての要素が0になるとき「ねじれなし」となります。
    その場合、基本ベクトルを他の基本ベクトルで偏微分したとき対称性が表れます。
  * 例えば、球面において、緯度に依存する回転を持つベクトル場は、ねじれを生じます。
  * Einsteinは、統一場理論のためにねじれを持つベクトル場を研究しました。
