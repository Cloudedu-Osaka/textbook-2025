# 第十二回: 接線と微分・積分

## 12.2 Descartesと放物線の接線

::::{admonition} ask ai
:class: tip

* Apolloniusは円錐曲線論で円錐曲線の直径や通径に付随して接線を定義しましたが、それは厳格な定義ではありませんでした

:::{admonition} ask ai repeatedly
:class: seealso

* 放物線については、直径と通径の再帰的な性質から、放物線と通径の交点における接線が直径上で交わることが示されました。
* 放物線がニ次多項式のグラフであることはMenaechmusによって発見されていましたが、
  高さは幅の自乗に比例する性質に基づいて、放物線の直径と通径から接線を作図することができました
* Archimedesはこの性質を用いて、放物線と通経に囲まれる領域の面積を積尽法で求めました
* Descartesは、座標空間において図形が代数方程式のグラフとして描画できることを発見し、
  円の接線が一点で交わるという性質を、連立方程式の判別式の値によって判定する手法を考案しました
* 放物線の内接円の中心は、頂点を通る直径上にありますが、
  放物線の任意の接線と軸の交点 (切片) と内接円の中心との中点は不動点であることが発見されます。
  これが放物線の焦点です
* Newtonは、放物線の軸に並行な光線が焦点に集光することに着目し、円錐曲線に一般化しました。
  焦点は、工学的な応用を持ちますが、現代の初等・中等教育では重視されていません
