## 12.1 Descartesと円の接線

::::{admonition} ask ai
:class: tip

* Euclidは原論で円の接線を次のように定義しました
  - A straight line is said to touch a circle which,
    meeting the circle and being produced, does not cut the circle.

:::{admonition} ask ai repeatedly
:class: seealso

* Euclidは、接線が直径と直交することを明確に証明できませんでした
  - The straight line drawn at right angles to the diameter of a circle
    from its end will fall outside the circle,
    and into the space between the straight line
    and the circumference another straight line cannot be interposed,
    further the angle of the semicircle is greater, 
    and the remaining angle less, than any acute rectilinear angle.
* Descartesは、連立方程式の判別式を用いて、円の直径に直交する直線 (通径) が一点で接する条件を示しました
* Descartesは、放物線の接線を、内接する円との共通な接線として定義しようとしました。
  その作図方法は、放物線の焦点の発見に繋がりました
