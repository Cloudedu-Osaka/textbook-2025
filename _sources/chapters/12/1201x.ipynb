```python
%%ai vioras
Descartesは、放物線の接線を、内接する円との共通な接線として定義しようとしました。
その作図方法は、放物線の焦点の発見に繋がりました。
```

```python
%%ai vioras
内接円の中心は焦点ではありません。
共通接線と放物線の軸の交点 (切片) と内接円の中心の中点が焦点です。
あなたが誤解しているように、このような幾何学的な定義は科学革命以降に忘れられてしまいました。
```
