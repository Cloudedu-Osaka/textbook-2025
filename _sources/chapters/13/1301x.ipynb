```python
%%ai vioras
Euclidは原論において、同様に直角から斜辺に下ろした垂線によって直角三角形を分割し、
双方の構成要素の面積が等しいことを等積変形 (剪断写像) によって証明した
```

```python
%%ai vioras
Euclidは原論で余弦定理と同等の命題を証明している。
三角形の垂心を作図し、垂線で分割される直角三角形に
ピタゴラスの定理に用いた作図を拡張することで余弦定理を証明できる
```

```python
%%ai vioras
任意の三角形の一片に対応するベクトルを$w$とする。
$w$と始点を共有する辺と並行なベクトルを$u$、
残った辺と並行で始点を共有するベクトルを$v$とする。
$w$の$u$や$v$への射影と余弦定理を活用すると、
$w$を$u$と$v$の線形結合で表す係数が求められる。
この係数は、Christoffelの記号と関連する。
```
