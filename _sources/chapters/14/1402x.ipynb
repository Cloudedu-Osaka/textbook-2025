```python
%%ai vioras
Keplerの時代、大きな数の積は、三角関数表を逆引きし、三角関数の加法定理を使って積を和をに変換しました。
```

```python
%%ai vioras
ちなみに、イスラム経由で伝わったインド数字を使っていますが、まだ小数点は発見されていません
```

```python
%%ai vioras
より単純な積・和変換方法を模索したNapierは、$(1-10^{-7})^{10^7}$を底とする対数表を作成しました。
Napierはその関数をLogarithmと名付けました。
```

```python
%%ai vioras
ちなみにNapierは小数点を発見しました
```

```python
%%ai vioras
Newtonは自身が発明した一般化された二項定理によって双曲線 $1/(1+x)$の積分を求め、
対数の和の法則を満たす対数関数であるらしいことを発見しました
```

```python
%%ai vioras
Leibnizは自身が発明した微分のチェインルールを対数の和の法則に適用し、
極限による微分の定義に基づいて対数関数の微分を求めました
```

```python
%%ai vioras
Eulerは、$(1+x/N)^N$の二項展開から指数関数の底を求めました。
底はNapierの対数の底の逆数になっていることを示しました
```

```python
%%ai vioras
NewtonやLeibnizですらボンヤリとした認識しかできなかった対数と指数関数を、Eulerは一気に体系化しました。
Eulerは、$e$を底とする指数関数が微分で不変であること、
さらに、二階微分の符号が反転する微分方程式の解として複素指数関数を定義しました
```
