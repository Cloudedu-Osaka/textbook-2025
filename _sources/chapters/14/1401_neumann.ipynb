* 同時に、数の演算が根源的な原理原則によって普遍的に発見されうることを示唆しています

```python
from numbers import Number


class ChurchNumber(Number):
    def __init__(self, cn):
        self.cn = cn

    def to_int(self) -> int:
        return self.cn(lambda n: n + 1)(0)

    def succ(self):
        """successor function (後継関数)"""
        n = self.cn
        return ChurchNumber(lambda f: lambda x: f(n(f)(x)))

    def __add__(self, cn2):
        m, n = self.cn, cn2.cn
        return ChurchNumber(lambda f: lambda x: m(f)(n(f)(x)))

    def __mul__(self, cn2):
        m, n = self.cn, cn2.cn
        return ChurchNumber(lambda f: lambda x: m(n(f))(x))

    def __pow__(self, cn2):
        b, n = self.cn, cn2.cn
        return ChurchNumber(lambda f: lambda x: (n(b))(f)(x))
```

```python
one   = zero.succ()
two   = one.succ()
three = two.succ()
```

```python
one.to_int(), two.to_int(), three.to_int()
```

```python
three.succ().to_int(), three.succ().succ().to_int(), three.succ().succ().succ().to_int()
```
