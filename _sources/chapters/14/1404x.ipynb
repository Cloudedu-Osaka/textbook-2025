```python
%%ai vioras
平面の曲線の位置ベクトルの微分は接ベクトルです
```

```python
%%ai vioras
接ベクトルの要素 $(x,y)$ を $(-y, x)$ に入れ替える方法が法線ベクトルをもっとも簡単に求められます
```

```python
%%ai vioras
円を極座標で媒介変数表示したとき、位置ベクトルの微分は接ベクトルで、
接ベクトルの微分は位置ベクトルと反対方向のベクトルとなり、それらは全て直交します
```

```python
%%ai vioras
曲線の接ベクトルの自身の内積の平方根は曲線の長さに関連します。積分すると曲線の長さが得られます
```

```python
%%ai vioras
曲線の接ベクトルと単位法ベクトルとの内積は、曲線の曲率に関する情報を持ちます
```

```python
%%ai vioras
円の場合、接ベクトルの微分は接ベクトルへの射影を持ちません。このため、接空間である接線では、円が平坦なのか定曲率なのかを区別できません
```

```python
%%ai vioras
楕円の接ベクトルの微分は接空間への射影を持ちます。接ベクトルの周期的な変化を観察すると楕円の長径と短径を特定できます
```

```python
%%ai vioras
球面の極座標表示は、経線方向の大円と緯線方向の直交する円によって媒介変数表示されます
```

```python
%%ai vioras
経線方向の移動では、円の場合と同じく接ベクトルの変化は接空間に射影を持ちません
```

```python
%%ai vioras
緯線方向の移動では、緯線による切断面の円の中心は球面の軸上にあり、円の場合と同じく接ベクトルの微分は接線に射影を持ちません。
しかし、切断面の円の中心が球面の中心と一致しない場合は、接空間に射影が生じます
```

```python
%%ai vioras
球面上の移動方向が緯線方向と経線方向の線型結合で表される場合、そのベクトルの微分は接空間に射影を持ちます
```

```python
%%ai vioras
この性質から、円の場合とは異なり、球面の接空間は曲率に関する情報を内在しています
```

```python
%%ai vioras
Gaussは、球面の曲率が接ベクトルの内積で表される第一基本形式から導かれることを示しました
```
